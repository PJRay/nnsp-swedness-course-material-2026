# Simple Powder Diffractometer testing
The classic `.instr` format can be converted with the McStasScript python API can make a widget suitable for e-learning. The translate tool is called `mcstas-pygen`, so for this example I called:
```
mcstas-pygen SimplePowderDiffractometer.instr
```
Which generated `SimplePowderDiffractometer_generated.py`, that can be directly imported into a notebook. 

You need to edit the file to change the parameter names and comments (the descriptions of these are not taken from the `.instr` automatically). There are different keyword arguments that can also be added for the parameters in here: for example you can use the `options` keyword to add a dropdown menu for one of the parameters.

Further documentation about widgets: https://panosc-vinyl.github.io/McStasScript/user_guide/widgets.html
